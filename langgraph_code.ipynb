```python
import os
from typing import TypedDict
from langgraph.graph import StateGraph,START,END
from langchain_google_genai import ChatGoogleGenerativeAI
!pip install -U langchain-google-genai
```

```python
os.environ["GOOGLE_API_KEY"] = "Enter API key here"
llm = ChatGoogleGenerativeAI(model="gemini-flash-lite-latest",timeout = 60,max_retries = 1)
```

```python

class OrderState(TypedDict):
    food : str

def cook_food(order: OrderState):
    prompt = f"You are a bakery owner.you have to convince user to buy the bakery product: {order['food']}"
    response = llm.invoke(prompt)
    return {"food":response.text}
    return {"food": "Cooked " + order["food"]}
def pack_food(order:OrderState):
    return {"food":order["food"] + " (Packed in Box)"}
# def trash_food(order:OrderState):
#     return {"food":order["food"].replace("Burnt","")}
# def check_quality(order:OrderState):
#     if "Burnt" in order["food"]:
#         return "trash"
#     return "pack"
    

builder = StateGraph(OrderState)
builder.add_node("cook",cook_food)
builder.add_node("pack",pack_food)
# builder.add_node("trash",trash_food)

builder.add_edge(START,"cook")
# builder.add_conditional_edges("cook",check_quality)
builder.add_edge("cook","pack")
builder.add_edge("pack",END)


kitchen = builder.compile()

print(kitchen.invoke({"food":" paneer Burger"}))
```
